Standard Library Overview


Python's Standard Library is a vast collection of modules and packages that come bundled with Python,
providing a wide range of functionalities out of the box. Here's an overview of some of the most commonly
used modules and packages in the Python Standard Library.


```python
import array
arr=array.array('i',[1,2,3,4,5])
print(arr)
```

```python
import math
print(math.sqrt(16))
print(math.pi)
```

```python
## random


import random
print(random.randint(1,50))
print(random.choice(['apple','banana','cherry','date']))
```

```python
## High level operations on files and collections of files
import shutil
shutil.copyfile('source.txt', 'destination.txt')  # Copy a file
```

```python
## Data Serialization
import json
data = {'name': 'John', 'age': 30, 'city': 'New York'}

json_str=json.dumps(data)  # Serialize to JSON string
print(json_str)
print(type(json_str))

parsed_data=json.loads(json_str)  # Deserialize back to Python object
print(parsed_data)
print(type(parsed_data))
```

```python
##CSV

import csv

with open('data.csv', mode='w', newline='') as file:
    writer=csv.writer(file)
    writer.writerow(['name','age'])
    writer.writerow(['stephen','32'])

with open('data.csv',mode='r') as file:
    reader=csv.reader(file)
    for row in reader:
        print(row)
```

```python
##datetime
from datetime import datetime,timedelta

now=datetime.now()
print("Current date and time:",now)

yesterday=now-timedelta(days=1)
print("Yesterday's date and time:",yesterday)
```

```python
## time
import time
print(time.time())  # Current time in seconds since epoch
time.sleep(2)  # Pause for 2 seconds
print(time.time())  # Current time in seconds since epoch
```

```python
##regular expressions
import re   

pattern = r'\d+'
text = 'There are 12 apples 45'
match = re.search(pattern, text)
print(match.group())  # Output: ['12', '34']
```
