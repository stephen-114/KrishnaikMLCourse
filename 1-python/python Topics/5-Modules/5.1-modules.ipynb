Importing Modeuls in python: Modules and Packages

in python,Modules and Package help organize and reuse code. heres a comprehension guide on how to import them


```python
import math
math.sqrt(19)
```

```python
from math import sqrt,pi
print(sqrt(16))
print(sqrt(25))
print(pi)
```

```python
import numpy as np
np.array([1,2,3,4])
```

```python
from math import *
```

```python
from math import *
print(sqrt(16))
print(pi)
```

```python
from package.maths import addition

addition(10,20)
```

```python
from package import maths
print(maths.addition(10,20))
```

```python
from package import maths
maths.subtraction(3-2)
```
