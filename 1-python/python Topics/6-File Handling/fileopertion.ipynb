File Operation- Read And Write Files


File handling is a crucial part of any programming language. Python provides built-in functions and methods to
read from and write to files, both text and binary. This lesson will cover the basics of file handling, including
reading and writing text files and binary files.

```python
## Read a Whole File

with open('example.txt','r') as file:
    content = file.read()
    print(content)
```

```python
## read a File line by line
with open('example.txt','r') as file:
    for line in file:
        print(line.strip()) ## strip() is used to remove the newline character at the end of each line
```

```python
## Wriinhg to a File
with open('example.txt','w') as file:
    file.write("This is a new line.\n")
    file.write("This is another new line.\n")
```

```python
##write a file without overwriting the existing content
with open('example.txt','a') as file:
    file.write("This line will be appended to the file.\n")
    file.write("This is another appended line.\n")
```

```python
##writing a lines to  a file
lines=['first line\n', 'second line\n', 'third line\n']
with open('example.txt','a') as file:
    file.writelines(lines)
```

```python
### Binary Files
data = b'\x00\x01\x02\x03\x04\x05'
with open('example.bin', 'wb') as file:
    file.write(data)
```

```python
## reading a binary file
with open('example.bin', 'rb') as file:
    content = file.read()
    print(content)
```

```python
### read the content from source text file and write it to a destination text file

with open('example.txt', 'r') as source_file:
    content = source_file.read()

with open('destination.txt', 'w') as destination_file:
    destination_file.write(content)
```

```python
## read a text file and write the number of lines,words,and characters.

def count_test_file(file_path):
    with open(file_path, 'r') as file:
        lines = file.readlines()
        lines_count=len(lines)
        word_count=sum(len(line.split()) for line in lines)
        char_count=sum(len(line) for line in lines)
    return lines_count, word_count, char_count

file_path = 'example.txt'
lines,words,characters=count_test_file(file_path)
print(f'lines: {lines}, words: {words}, characters: {characters}')

        
```

```python
## writing and then reading a file

with open('example.txt', 'w+') as file:
    file.write("This is a new line.\n")
    file.write("This is another new line.\n")
    
    file.seek(0)  # Move the cursor to the beginning of the file
    content = file.read()
    print(content)
```

```python
##checking of a path is a file or directory
import os
path = 'package'
if os.path.isfile(path):
    print(f"The path '{path}' is a file.") 
elif os.path.isdir(path):
    print(f"The path '{path}' is a directory.")
else:
    print(f"The path '{path}' is neither a file nor a directory.")
```
