Pandas-DataFrame and Series

Pandas is an open-source Python library used for data manipulation, analysis, and cleaning. It provides fast, flexible, and easy-to-use data structures like Series and DataFrame for working with structured data.

```python
import pandas as pd
data=[1,2,3,4,5]
series=pd.Series(data)
print(" series \n",series)
print(type(series))
```

```python
##create a series from a dictionary
data={'a':1,'b':2,'c':3}
series_dict=pd.Series(data)
print(series_dict)
```

```python
data=[10,20,30,40,50]
index=['a','b','c','d','e']
series=pd.Series(data,index=index)
print(series)
```

```python
## dataframe
## create a Dataframe from dictionary of list
data={
    'name':['stephen','siva','sreedhar'],
    'age':[25,30,35],
    'city':['New York','Los Angeles','Chicago']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
import numpy as np
np.array(df)
print(df)
```
