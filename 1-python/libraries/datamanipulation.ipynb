```python
## Renaming Columns
df=df.rename(columns={'Sales_Amount':'Sales_Cost'})
df.head()
```

```python
## Change DataTypes
df['Value_new']=df['Quantity_Sold'].fillna(df['Quantity_value'].mean()).astype(int)
df.head()
```

```python
df['new_sales_Cost']=df['Sales_Cost'].apply(lambda x:x**2)
df
```

```python
df.drop['Sale_Date']
```

```python
## Data Aggregating and grouping

grouped_mean=df.groupby('new_sales_Cost')['Sales_Cost'].mean()
```

### merging and joining dataframes 

```python
df1=pd.DataFrame({'key':['A','B','C','D','E'],'value':[10,20,30,40,50]})
df2=pd.DataFrame({'key':['A','C','E','F','G'],'value':[100,200,300,400,500]})
                         

```

```python
df1
```

```python
## merge df on the 'key_columns'
pd.merge(df1,df2,on="key",how='outer')
```
